# V1 · exploratory data analysis
Adapted from the CampusX Quora Question Pairs tutorial. Inspect training labels and
repeated questions after building the full question graph. Use the same data and seed
as the maintained V1 training entry point.

```python
from pathlib import Path
import sys

root = next(folder for folder in (Path.cwd(), *Path.cwd().parents)
            if (folder / 'v1' / 'baseline.py').is_file())
if str(root) not in sys.path:
    sys.path.insert(0, str(root))

from v2.data import load_pairs, question_groups, split_pairs

pairs = load_pairs(root / 'data' / 'train.csv')
pairs['group'] = question_groups(pairs)
usable = pairs[pairs.clean1.ne('') & pairs.clean2.ne('')]
splits = split_pairs(usable, seed=42)
{key: len(part) for key, part in splits.items()}
```

```python
train = splits['train']
train['is_duplicate'].value_counts().rename(index={0: 'distinct', 1: 'duplicate'})
```

```python
import pandas as pd

frequency = pd.concat([train.qid1, train.qid2]).value_counts()
frequency.describe()
```
